```python
import cv2
import matplotlib.pyplot as plt
import urllib.request
import numpy as np

# 1. Test için örnek renkli bir görüntü indirelim
url = 'https://raw.githubusercontent.com/opencv/opencv/master/samples/data/lena.jpg'
req = urllib.request.urlopen(url)
arr = np.asarray(bytearray(req.read()), dtype=np.uint8)
img = cv2.imdecode(arr, cv2.IMREAD_COLOR)

# 2. Sıfırdan Kendi calcHist Fonksiyonumuz
def custom_calc_hist(channel):
    """
    Bir görüntü kanalındaki 0-255 arası piksel değerlerinin
    frekansını (sayısını) sıfırdan hesaplar.
    """
    # Görüntüyü tek boyutlu bir diziye (vektöre) çeviriyoruz
    flattened = channel.ravel()

    # 0 ile 255 arasındaki her bir değerin kaç kez geçtiğini bincount ile sayıyoruz
    hist = np.bincount(flattened, minlength=256)

    # OpenCV formatına uygun olması için (256, 1) boyutuna getiriyoruz
    return hist.reshape(256, 1)

# 3. Kanalları Ayırma (BGR)
color = ('b', 'g', 'r')
channels = cv2.split(img)

# 4. Histogramları Çizdirme
plt.figure(figsize=(10, 5))

# Orijinal resmi görmek için yanına koyalım
plt.subplot(1, 2, 1)
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.title('Orijinal Renkli Görüntü')
plt.axis('off')

plt.subplot(1, 2, 2)
for channel, col in zip(channels, color):
    # Kendi yazdığımız fonksiyonu her kanal için ayrı ayrı çalıştırıyoruz
    hist = custom_calc_hist(channel)
    plt.plot(hist, color=col, label=f'{col.upper()} Kanalı')
    plt.xlim([0, 256])

plt.title('Özel calcHist() ile Renk Kanalları')
plt.xlabel('Piksel Değeri (0-255)')
plt.ylabel('Piksel Sayısı')
plt.legend()
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```
