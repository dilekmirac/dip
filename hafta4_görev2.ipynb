```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
import urllib.request

# 1. Test için örnek bir görüntü indirelim
url = 'https://picsum.photos/id/237/400/400'
req = urllib.request.urlopen(url)
arr = np.asarray(bytearray(req.read()), dtype=np.uint8)
img = cv2.imdecode(arr, cv2.IMREAD_GRAYSCALE)

# 2. Gauss Filtresi ile Görüntüyü Hafif Yumuşatma
blurred_img = cv2.GaussianBlur(img, (5, 5), 0)

# 3. Belirttiğin 3x3 Prewitt Filtresi (Dikey türev / Yatay kenar tespiti)
prewitt_kernel_vertical = np.array([
    [-1, -1, -1],
    [ 0,  0,  0],
    [ 1,  1,  1]
], dtype=np.float32)

# 4. Filtreyi görüntü üzerinde uygulama
filtered_img = cv2.filter2D(blurred_img, -1, prewitt_kernel_vertical)

# 5. Sonuçları Görselleştirme
plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
plt.imshow(img, cmap='gray')
plt.title('1. Orijinal Gri Görüntü')
plt.axis('off')

plt.subplot(1, 3, 2)
plt.imshow(blurred_img, cmap='gray')
plt.title('2. Gauss Bulanıklaştırılmış')
plt.axis('off')

plt.subplot(1, 3, 3)
plt.imshow(filtered_img, cmap='gray')
plt.title('3. Prewitt (Yatay Kenar) Uygulanmış')
plt.axis('off')

plt.tight_layout()
plt.show()
```
