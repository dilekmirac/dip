## Görüntü İşlemede Kullanılan Renk Uzayları

### 1. RGB / BGR (Red, Green, Blue / Blue, Green, Red)
* **Avantajları:** Ekranların, kameraların ve temel görüntü işleme kütüphanelerinin doğal çalışma yapısıdır; kanal okuma ve doğrudan görselleştirme işlemleri pratiktir.
* **Dezavantajları:** Parlaklık (aydınlatma şiddeti) ile renk bilgisi birbirine karışıktır; ortam ışığındaki değişimler tüm kanalları birlikte etkiler.
* **Tercih Edildiği Problemler:** Görüntü okuma/kaydetme, ekran gösterimi, temel piksel ve kanal manipülasyonları.

### 2. Grayscale (Gri Tonlama)
* **Avantajları:** Tek kanal üzerinden çalışıldığı için bellek kullanımı ve işlem maliyeti çok düşüktür; yoğunluk/parlaklık ve yapısal geçişlere odaklanmayı kolaylaştırır.
* **Dezavantajları:** Renk bilgisini tamamen ortadan kaldırır; renk farkına dayalı ayrımlarda yetersiz kalır.
* **Tercih Edildiği Problemler:** Kenar tespiti, eşikleme, morfolojik işlemler, kontur/şekil analizi ve yüksek hız gerektiren ön işlemler.

### 3. HSV / HSI (Hue, Saturation, Value / Intensity)
* **Avantajları:** Renk tonunu (*Hue*), doygunluğu (*Saturation*) ve parlaklığı (*Value/Intensity*) birbirinden kısmen ayırır; renk aralıklarıyla filtreleme yaparken insan algısına daha yakındır.
* **Dezavantajları:** Doygunluk veya parlaklık çok düştüğünde renk tonu değeri kararsızlaşır; ton ekseni dairesel olduğu için sınır değerlerde (örneğin kırmızı tonun uçlarında) eşik ayarı dikkat gerektirir.
* **Tercih Edildiği Problemler:** Belirli renkteki nesnelerin tespiti ve takibi, renk tabanlı maskeleme ve değişen ışık koşullarında nesne ayrımı.

### 4. YUV / YCbCr
* **Avantajları:** Parlaklık bileşenini (*Y*) renk farkı bileşenlerinden (*U/V* veya *Cb/Cr*) ayırır; insan gözünün parlaklıktaki değişimlere renk değişimlerinden daha duyarlı olması prensibinden yararlanır.
* **Dezavantajları:** Doğrudan RGB sensör verisine göre dönüşüm adımı gerektirir; belirli uç aydınlatmalarda eşik değerlerinin kalibrasyonu gerekebilir.
* **Tercih Edildiği Problemler:** Video sıkıştırma ve aktarım sistemleri, ten rengi tespiti ve parlaklık-renk ayrımının istendiği uygulamalar.

### 5. CIE Lab (L\*, a\*, b\*)
* **Avantajları:** Algısal olarak daha homojen bir yapı sunar; iki renk arasındaki sayısal fark, insan gözünün algıladığı renk farkına daha yakındır.
* **Dezavantajları:** Dönüşüm hesabı RGB ve HSV'ye kıyasla daha maliyetlidir; referans beyaz noktasına bağlıdır.
* **Tercih Edildiği Problemler:** Hassas renk farkı ölçümü, renk kalibrasyonu, endüstriyel kalite kontrol ve görsel benzerlik değerlendirmeleri.

### Özet Seçim Rehberi
* **Hız ve kenar/yapı analizi:** Grayscale
* **Renk filtreleme ve nesne takibi:** HSV veya YCbCr
* **Video iletimi ve parlaklık-renk ayrımı:** YUV / YCbCr
* **Algısal renk farkı ve kalite kontrol:** CIE Lab
